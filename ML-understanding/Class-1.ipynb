```python
df_clean['sex'] = df_clean['sex'].map({'male': 0, 'female': 1})#label incoding to conver yes no or male female into 0-1 numric
```

```python
df_clean.rename(columns={
    'sex': 'is_female',
    'smoker': 'is_smoker'
    
},inplace= True)
```
